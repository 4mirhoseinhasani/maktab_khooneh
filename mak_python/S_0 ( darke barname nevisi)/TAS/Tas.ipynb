# Sakht yek list 6 ozvi be onvan adade Tas

```python
Tas = [0, 0, 0, 0, 0, 0]
```

```python
import random
random.seed()
```

# save dar list
## halghe baraye tekrar roll

```python
for bar in range(0, 15000):
    roll = random.randint(1, 6)
    Tas[roll - 1] = Tas[roll - 1] + 1
```

# make chart for project!

```python
import matplotlib.pyplot as plt
%matplotlib inline
numbers = [1, 2, 3, 4, 5, 6]
plt.bar(numbers, Tas, color = 'red')
plt.xlabel('Tas')
plt.ylabel('roll')
plt.show()
```
